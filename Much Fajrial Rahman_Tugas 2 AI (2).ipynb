### **PROMPT 1 : PENGOLAHAN DATA SEISMIK REFRAKSI**

```python
# ============================================================
# METODE 1: SEISMIK REFLEKSI
# QC WELL-TO-SEISMIC TIE (MD-TWT)
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Folder output
output_dir = "hasil_seismik"
os.makedirs(output_dir, exist_ok=True)

# 2. Input data MD-TWT Well Poseidon-1
data = {
    "MD_m": [
        0.0, 492.0, 1500.0, 2300.0, 3120.0,
        4250.0, 4615.0, 4796.0, 4950.0, 5112.0
    ],
    "TWT_ms": [
        0.0, 656.0, 1620.0, 2180.0, 2685.0,
        3140.0, 3295.0, 3388.0, 3472.0, 3560.0
    ]
}

df = pd.DataFrame(data)

# 3. Pemeriksaan kualitas data
print("=== QUALITY CONTROL SEISMIK ===")
print("Jumlah data:", len(df))
print("\nData kosong:\n", df.isnull().sum())

if not df["MD_m"].is_monotonic_increasing:
    raise ValueError("MD tidak meningkat secara monoton.")

if not df["TWT_ms"].is_monotonic_increasing:
    raise ValueError("TWT tidak meningkat secara monoton.")

if df["MD_m"].duplicated().any():
    raise ValueError("Terdapat MD duplikat.")

# 4. Menghitung perubahan kedalaman dan waktu
df["Delta_MD_m"] = df["MD_m"].diff()
df["Delta_TWT_ms"] = df["TWT_ms"].diff()

# 5. Menghitung kecepatan interval
# V = 2 * perubahan kedalaman / perubahan TWT
# TWT dalam detik

df["Interval_Velocity_m_s"] = (
    2 * df["Delta_MD_m"] /
    (df["Delta_TWT_ms"] / 1000)
)

df.loc[
    df["Delta_TWT_ms"] <= 0,
    "Interval_Velocity_m_s"
] = np.nan

print("\n=== HASIL PERHITUNGAN ===")
display(df)

# 6. Statistik data
print("\nRentang MD:", df["MD_m"].min(), "-", df["MD_m"].max(), "m")
print("Rentang TWT:", df["TWT_ms"].min(), "-", df["TWT_ms"].max(), "ms")

# 7. Plot MD-TWT
plt.figure(figsize=(9, 6))

plt.plot(
    df["TWT_ms"],
    df["MD_m"],
    marker="o",
    linestyle="-",
    label="Poseidon-1"
)

plt.xlabel("Two-Way Travel Time (ms)")
plt.ylabel("Measured Depth (m)")
plt.title("Well Time-Depth Relationship - Poseidon-1")

plt.gca().invert_yaxis()
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()

plt.savefig(
    os.path.join(output_dir, "seismik_md_twt.png"),
    dpi=300
)

plt.show()

# 8. Plot kecepatan interval
plt.figure(figsize=(8, 6))

plt.plot(
    df["Interval_Velocity_m_s"],
    df["MD_m"],
    marker="o"
)

plt.xlabel("Kecepatan Interval (m/s)")
plt.ylabel("Measured Depth (m)")
plt.title("Interval Velocity - Poseidon-1")

plt.gca().invert_yaxis()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    os.path.join(output_dir, "seismik_interval_velocity.png"),
    dpi=300
)

plt.show()

# 9. Simpan hasil
df.to_csv(
    os.path.join(output_dir, "hasil_well_tie.csv"),
    index=False
)

print("\nPengolahan selesai.")
print("Output:", os.path.abspath(output_dir))
```

### **PROMPT 3 : Geomegnetik**

```python
# ============================================================
# METODE 3: GEOMAGNETIK
# ANALISIS ANOMALI MAGNETIK RELATIF
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.interpolate import griddata

# 1. Folder output
output_dir = "hasil_geomagnetik"
os.makedirs(output_dir, exist_ok=True)

# 2. Data sintetis untuk demonstrasi
data = {
    "X_m": [0, 100, 200, 300, 400, 0, 100, 200, 300, 400],
    "Y_m": [0, 0, 0, 0, 0, 100, 100, 100, 100, 100],
    "T_obs_nT": [
        45000, 45012, 45025, 45010, 44995,
        45005, 45020, 45035, 45015, 45000
    ]
}

df = pd.DataFrame(data)

# 3. Quality control
print("=== QUALITY CONTROL GEOMAGNETIK ===")
print("Jumlah titik:", len(df))

if df[["X_m", "Y_m", "T_obs_nT"]].isnull().any().any():
    raise ValueError("Terdapat data koordinat atau magnetik yang kosong.")

# 4. Menghitung anomali magnetik relatif
# Catatan: bukan pengganti koreksi IGRF dan diurnal.

df["Anomali_Relatif_nT"] = (
    df["T_obs_nT"] - df["T_obs_nT"].mean()
)

print("\n=== STATISTIK ANOMALI ===")
print("Minimum:", df["Anomali_Relatif_nT"].min(), "nT")
print("Maksimum:", df["Anomali_Relatif_nT"].max(), "nT")
print("Rata-rata:", df["Anomali_Relatif_nT"].mean(), "nT")

display(df)

# 5. Interpolasi untuk peta
x = df["X_m"].to_numpy()
y = df["Y_m"].to_numpy()
z = df["Anomali_Relatif_nT"].to_numpy()

xi = np.linspace(x.min(), x.max(), 100)
yi = np.linspace(y.min(), y.max(), 100)

XI, YI = np.meshgrid(xi, yi)

ZI = griddata(
    (x, y),
    z,
    (XI, YI),
    method="linear"
)

# 6. Membuat peta kontur
plt.figure(figsize=(9, 6))

contour = plt.contourf(
    XI,
    YI,
    ZI,
    levels=20
)

plt.scatter(
    x,
    y,
    marker="o",
    edgecolors="black"
)

plt.colorbar(
    contour,
    label="Anomali Magnetik Relatif (nT)"
)

plt.xlabel("Koordinat X (m)")
plt.ylabel("Koordinat Y (m)")
plt.title("Peta Anomali Magnetik Relatif")

plt.axis("equal")
plt.tight_layout()

plt.savefig(
    os.path.join(output_dir, "peta_geomagnetik.png"),
    dpi=300
)

plt.show()

# 7. Grafik anomali berdasarkan nomor titik
plt.figure(figsize=(9, 5))

plt.plot(
    np.arange(1, len(df) + 1),
    df["Anomali_Relatif_nT"],
    marker="o"
)

plt.xlabel("Nomor Titik Pengukuran")
plt.ylabel("Anomali Magnetik Relatif (nT)")
plt.title("Profil Anomali Magnetik")

plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    os.path.join(output_dir, "profil_geomagnetik.png"),
    dpi=300
)

plt.show()

# 8. Simpan hasil
df.to_csv(
    os.path.join(output_dir, "hasil_geomagnetik.csv"),
    index=False
)

print("\nPengolahan selesai.")
print("Output:", os.path.abspath(output_dir))
```

### **PROMPT 5 : Magnetotelurik**

```python
# ============================================================
# METODE 5: MAGNETOTELURIK (MT)
# QC DAN VISUALISASI RESISTIVITAS SEMU SERTA FASE
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Folder output
output_dir = "hasil_mt"
os.makedirs(output_dir, exist_ok=True)

# 2. Data sintetis untuk demonstrasi
data = {
    "Station": [
        "MT01", "MT01", "MT01", "MT01",
        "MT02", "MT02", "MT02", "MT02"
    ],
    "Frequency_Hz": [
        100, 10, 1, 0.1,
        100, 10, 1, 0.1
    ],
    "Rhoa_ohm_m": [
        20, 35, 60, 100,
        25, 40, 70, 120
    ],
    "Phase_deg": [
        35, 40, 45, 50,
        33, 39, 44, 48
    ]
}

df = pd.DataFrame(data)

# 3. Quality control
print("=== QUALITY CONTROL MT ===")
print("Jumlah data:", len(df))

df["QC_Status"] = "OK"

invalid = (
    df[["Frequency_Hz", "Rhoa_ohm_m", "Phase_deg"]].isnull().any(axis=1)
    | (df["Frequency_Hz"] <= 0)
    | (df["Rhoa_ohm_m"] <= 0)
)

df.loc[invalid, "QC_Status"] = "CHECK_VALUE"

valid = df[df["QC_Status"] == "OK"].copy()

if valid.empty:
    raise ValueError("Tidak terdapat data MT yang valid.")

# 4. Statistik data
print("\n=== STATISTIK MT ===")
print("Jumlah stasiun:", valid["Station"].nunique())
print("Rentang frekuensi:", valid["Frequency_Hz"].min(),
      "-", valid["Frequency_Hz"].max(), "Hz")
print("Rentang resistivitas semu:", valid["Rhoa_ohm_m"].min(),
      "-", valid["Rhoa_ohm_m"].max(), "ohm m")
print("Rentang fase:", valid["Phase_deg"].min(),
      "-", valid["Phase_deg"].max(), "derajat")

display(df)

# 5. Plot resistivitas semu terhadap frekuensi
plt.figure(figsize=(9, 6))

for station in valid["Station"].unique():
    subset = valid[valid["Station"] == station].sort_values(
        "Frequency_Hz"
    )

    plt.loglog(
        subset["Frequency_Hz"],
        subset["Rhoa_ohm_m"],
        marker="o",
        label=station
    )

plt.xlabel("Frekuensi (Hz)")
plt.ylabel("Resistivitas Semu (ohm m)")
plt.title("Kurva Resistivitas Semu Magnetotelurik")

plt.grid(True, which="both", alpha=0.3)
plt.legend()
plt.tight_layout()

plt.savefig(
    os.path.join(output_dir, "mt_resistivitas_semu.png"),
    dpi=300
)

plt.show()

# 6. Plot fase terhadap frekuensi
plt.figure(figsize=(9, 6))

for station in valid["Station"].unique():
    subset = valid[valid["Station"] == station].sort_values(
        "Frequency_Hz"
    )

    plt.semilogx(
        subset["Frequency_Hz"],
        subset["Phase_deg"],
        marker="o",
        label=station
    )

plt.xlabel("Frekuensi (Hz)")
plt.ylabel("Fase (derajat)")
plt.title("Kurva Fase Magnetotelurik")

plt.grid(True, which="both", alpha=0.3)
plt.legend()
plt.tight_layout()

plt.savefig(
    os.path.join(output_dir, "mt_fase.png"),
    dpi=300
)

plt.show()

# 7. Simpan hasil
df.to_csv(
    os.path.join(output_dir, "hasil_mt.csv"),
    index=False
)

print("\nPengolahan selesai.")
print("Output:", os.path.abspath(output_dir))
```
